```python
# Coaching Database
# Format: { student_id: {"name": ..., "father": ..., "village": ..., "phone": ..., "fee_amount": ..., "paid_months": []} }
coaching_db = {}
next_id = 1001  # Auto-generating ID starting from 1001

while True:
    print("\n================== COACHING MANAGEMENT SYSTEM ==================")
    print("1. New Admission (Naya Baccha)")
    print("2. Fee Deposit (Fees Jama Karein)")
    print("3. Student Passbook / Profile (Pura Detail)")
    print("4. Defaulter List (Fees Baaki Wali List)")
    print("5. Exit (System Band)")
    print("================================================================")
    print(" so you next one and giving around and wake now ")
    
    main_choice = input("Option chuniye (1/2/3/4/5): ").strip()

    if main_choice == "5":
        print("\nSystem band ho raha hai. Bye! Thank you!")
        break

    elif main_choice == "1":
        print("\n--- NEW ADMISSION ---")
        name = input("Enter Student's First Name: ").strip().lower()
        father_name = input("Enter Father's Name: ").strip().lower()
        village = input("Enter Village Name: ").strip().lower()
        phone = input("Enter 10-digit Phone Number: ").strip()
        monthly_fee = float(input("Enter Monthly Fee Amount: ₹"))
        
        # Auto generating Roll Number/ID
        student_id = str(next_id)
        next_id += 1
        
        # Saving details to database
        coaching_db[student_id] = {
            "name": name,
            "father": father_name,
            "village": village,
            "phone": phone,
            "fee_amount": monthly_fee,
            "paid_months": []  # Isme jama kiye hue mahino ke naam aayenge
        }
        
        print(f"\n✅ Success: Admission Ho Gaya!")
        print(f"🆔 Student ID / Roll No: {student_id}")
        print(f"👤 Name: {name.capitalize()}")

    elif main_choice == "2":
        print("\n--- FEE DEPOSIT ---")
        search_id = input("Enter Student ID / Roll No: ").strip()
        
        if search_id in coaching_db:
            student = coaching_db[search_id]
            print(f"\nStudent Found: {student['name'].capitalize()} s/o {student['father'].capitalize()}")
            print(f"Monthly Fee Rate: ₹{student['fee_amount']}")
            
            month_name = input("Kis Month ki fees jama karni hai? (e.g., May, June): ").strip().capitalize()
            
            # Check agar us month ki fees pehle se jama toh nahi hai
            if month_name in student["paid_months"]:
                print(f"⚠️ Alert: {month_name} ki fees pehle hi jama ho chuki hai!")
            else:
                amount_paid = float(input(f"Enter amount to pay (₹{student['fee_amount']}): ₹"))
                
                if amount_paid >= student['fee_amount']:
                    student["paid_months"].append(month_name)
                    print(f"✅ Success: {month_name} ki fee jama ho gayi!")
                else:
                    print("❌ Error: Puri monthly fee jama kijiye. Partial payment allowed nahi hai.")
        else:
            print("❌ Error: Yeh Student ID database me nahi hai!")

    elif main_choice == "3":
        print("\n--- STUDENT PASSBOOK & PROFILE ---")
        search_id = input("Enter Student ID / Roll No: ").strip()
        
        if search_id in coaching_db:
            st = coaching_db[search_id]
            print("\n======================================")
            print("         COACHING OFFICIAL PASSBOOK   ")
            print("======================================")
            print(f"Roll No / ID   : {search_id}")
            print(f"Student Name   : {st['name'].upper()}")
            print(f"Father's Name  : {st['father'].upper()}")
            print(f"Village Name   : {st['village'].upper()}")
            print(f"Phone Number   : {st['phone']}")
            print(f"Monthly Fee    : ₹{st['fee_amount']}")
            print("--------------------------------------")
            print("          PAID MONTHS RECORD          ")
            
            if len(st["paid_months"]) == 0:
                print("Koi fee jama nahi hui hai abhi tak.")
            else:
                for month in st["paid_months"]:
                    print(f"✔️ Fee Paid for Month: {month}")
                    
            print("======================================")
        else:
            print("❌ Error: Student ID nahi mili!")

    elif main_choice == "4":
        print("\n--- DEFAULTER LIST (FEES PENDING) ---")
        check_month = input("Kis month ke defaulters dekhne hain? (e.g., May, June): ").strip().capitalize()
        
        print(f"\n--- Students who HAVEN'T paid for {check_month} ---")
        print(f"{'ID':<6} | {'Student Name':<15} | {'Father Name':<15} | {'Phone':<12}")
        print("-" * 55)
        
        found_defaulter = False
        for s_id, data in coaching_db.items():
            if check_month not in data["paid_months"]:
                print(f"{s_id:<6} | {data['name'].capitalize():<15} | {data['father'].capitalize():<15} | {data['phone']:<12}")
                found_defaulter = True
                
        if not found_defaulter:
            print(f"Wow! Sabhi bacchon ne {check_month} ki fees de di hai.")
            
    else:
        print("Galat option! Sahi number chuniye (1-5).")
```

```python
import json
import os
from datetime import datetime

DATA_FILE = "coaching_data.json"

# 1. DATA LOAD AUR SAVE KARNE KE FUNCTIONS
def load_data():
    """File se data load karne ke liye"""
    if os.path.exists(DATA_FILE):
        with open(DATA_FILE, "r") as file:
            try:
                data = json.load(file)
                # Next ID nikalne ke liye
                if data:
                    all_ids = [int(i) for i in data.keys()]
                    next_id = max(all_ids) + 1
                else:
                    next_id = 1001
                return data, next_id
            except json.JSONDecodeError:
                return {}, 1001
    return {}, 1001

def save_data(data):
    """Data ko permanent save karne ke liye"""
    with open(DATA_FILE, "w") as file:
        json.dump(data, file, indent=4)

# Data load karein
coaching_db, next_id = load_data()

# Current month automatic nikalne ke liye (jaise: May, June)
current_month = datetime.now().strftime("%B")

while True:
    print("\n================== ADVANCED COACHING SYSTEM ==================")
    print(f"Current Month: {current_month}")
    print("1. New Admission (Naya Baccha + Survey Detail)")
    print("2. Fee Deposit & Print Parchi (Receipt)")
    print("3. Student Login / Search (By ID or Name)")
    print("4. Defaulter List (Fees Baaki Wali List)")
    print("5. Save & Exit (System Band)")
    print("==============================================================")
    
    main_choice = input("Option chuniye (1-5): ").strip()

    if main_choice == "5":
        save_data(coaching_db)
        print("\n✅ Saara data hamesha ke liye save ho gaya hai. System band ho raha hai. Bye!")
        break

    elif main_choice == "1":
        print("\n--- NEW ADMISSION & SURVEY ---")
        name = input("Enter Student's First Name: ").strip().lower()
        father_name = input("Enter Father's Name: ").strip().lower()
        village = input("Enter Village Name: ").strip().lower()
        phone = input("Enter 10-digit Phone Number: ").strip()
        monthly_fee = float(input("Enter Monthly Fee Amount: ₹"))
        
        # Survey Data Collection for Improvement
        print("\n--- Student Survey Data ---")
        prev_coaching = input("Pehle kisi coaching me padhte the? (Yes/No): ").strip()
        weak_subject = input("Sabse weak subject kaun sa hai?: ").strip()
        student_goal = input("Bacche ka target/goal kya hai? (Board Exam/Job/Etc): ").strip()
        
        student_id = str(next_id)
        next_id += 1
        
        # Database me save karna
        coaching_db[student_id] = {
            "name": name,
            "father": father_name,
            "village": village,
            "phone": phone,
            "fee_amount": monthly_fee,
            "paid_months": [],
            "survey": {
                "prev_coaching": prev_coaching,
                "weak_subject": weak_subject,
                "goal": student_goal
            }
        }
        save_data(coaching_db) # Turant save karein
        print(f"\n✅ Admission Successful! Roll No / ID generated: {student_id}")

    elif main_choice == "2":
        print("\n--- FEE DEPOSIT & RECEIPT ---")
        search_id = input("Enter Student ID / Roll No: ").strip()
        
        if search_id in coaching_db:
            student = coaching_db[search_id]
            print(f"\nBaccha Mila: {student['name'].capitalize()} (ID: {search_id})")
            month_name = input(f"Kis Month ki fees jama karni hai? (Default {current_month}): ").strip().capitalize()
            if not month_name:
                month_name = current_month
                
            if month_name in student["paid_months"]:
                print(f"⚠️ Alert: {month_name} ki fees pehle hi jama hai!")
            else:
                amount_paid = float(input(f"Enter Fee Amount (₹{student['fee_amount']}): ₹"))
                if amount_paid >= student['fee_amount']:
                    student["paid_months"].append(month_name)
                    save_data(coaching_db)
                    
                    # PARCHI (TXT FILE) GENERATE KARNA
                    receipt_filename = f"Receipt_{search_id}_{month_name}.txt"
                    with open(receipt_filename, "w") as receipt:
                        receipt.write("======================================\n")
                        receipt.write("         COACHING FEE RECEIPT         \n")
                        receipt.write("======================================\n")
                        receipt.write(f"Receipt Date : {datetime.now().strftime('%d-%m-%Y %H:%M')}\n")
                        receipt.write(f"Roll No / ID : {search_id}\n")
                        receipt.write(f"Student Name : {student['name'].upper()}\n")
                        receipt.write(f"Father's Name: {student['father'].upper()}\n")
                        receipt.write(f"Paid Month   : {month_name.upper()}\n")
                        receipt.write(f"Amount Paid  : ₹{amount_paid}\n")
                        receipt.write("Status       : PAID & CLEARED\n")
                        receipt.write("======================================\n")
                        receipt.write("       Thank you for the payment!      \n")
                    
                    print(f"✅ Fee jama ho gayi! Parchi file '{receipt_filename}' ban gayi hai. Aap isse print kar sakte hain.")
                else:
                    print("❌ Error: Puri fee jama kijiye!")
        else:
            print("❌ Error: ID nahi mili!")

    elif main_choice == "3":
        print("\n--- STUDENT LOGIN & SEARCH ---")
        search_type = input("ID se search karna hai ya Name se? (id/name): ").strip().lower()
        found_student_id = None
        
        if search_type == "id":
            s_id = input("Enter Student ID: ").strip()
            if s_id in coaching_db:
                found_student_id = s_id
        elif search_type == "name":
            s_name = input("Enter Student Name: ").strip().lower()
            for k, v in coaching_db.items():
                if v["name"] == s_name:
                    found_student_id = k
                    break
                    
        if found_student_id:
            st = coaching_db[found_student_id]
            print("\n======================================")
            print("         STUDENT OFFICIAL PROFILE     ")
            print("======================================")
            print(f"ID / Roll No   : {found_student_id}")
            print(f"Student Name   : {st['name'].upper()}")
            print(f"Father's Name  : {st['father'].upper()}")
            print(f"Village Name   : {st['village'].upper()}")
            print(f"Phone Number   : {st['phone']}")
            print(f"Monthly Fee    : ₹{st['fee_amount']}")
            print("--------------------------------------")
            
            # AUTOMATIC REMINDER CHECK
            if current_month in st["paid_months"]:
                print(f"Fee Status     : ✅ CURRENT MONTH ({current_month}) PAID")
            else:
                print(f"Fee Status     : ⚠️ REMINDER: {current_month} FEE IS PENDING!")
                
            print("--------------------------------------")
            print("          PAID MONTHS RECORD          ")
            print(f"Paid Months    : {', '.join(st['paid_months']) if st['paid_months'] else 'No records'}")
            print("--------------------------------------")
            print("          SURVEY & IMPROVEMENT DATA   ")
            print(f"Weak Subject   : {st['survey']['weak_subject']}")
            print(f"Student Goal   : {st['survey']['goal']}")
            print(f"Prev Coaching  : {st['survey']['prev_coaching']}")
            print("======================================")
        else:
            print("❌ Error: Student ki koi detail nahi mili!")

    elif main_choice == "4":
        print("\n--- DEFAULTER LIST ---")
        month_to_check = input(f"Kis month ke defaulters dekhne hain? (Default {current_month}): ").strip().capitalize()
        if not month_to_check:
            month_to_check = current_month
            
        print(f"\n--- Students with PENDING fees for {month_to_check} ---")
        print(f"{'ID':<6} | {'Student Name':<15} | {'Phone':<12} | {'Village':<15}")
        print("-" * 55)
        
        has_defaulters = False
        for s_id, data in coaching_db.items():
            if month_to_check not in data["paid_months"]:
                print(f"{s_id:<6} | {data['name'].capitalize():<15} | {data['phone']:<12} | {data['village'].capitalize():<15}")
                has_defaulters = True
                
        if not has_defaulters:
            print(f"Bahut badhiya! {month_to_check} me sabki fees jama hai.")
            
    else:
        print("Galat option! Sahi number chuniye.")
```

```python
import json
import os
import pandas as pd

SURVEY_FILE = "coaching_survey_data.json"
SURVEY_EXCEL = "Coaching_Survey_Report.xlsx"

def load_survey_data():
    """File se data load karne aur agla ID number nikalne ke liye"""
    if os.path.exists(SURVEY_FILE):
        with open(SURVEY_FILE, "r") as file:
            try:
                data = json.load(file)
                if data:
                    # Saari numeric IDs nikal kar max nikalna
                    all_ids = [int(i.split("-")[1]) for i in data.keys()]
                    next_id = max(all_ids) + 1
                else:
                    next_id = 1001
                return data, next_id
            except json.JSONDecodeError:
                return {}, 1001
    return {}, 1001

def save_survey_data(data):
    with open(SURVEY_FILE, "w") as file:
        json.dump(data, file, indent=4)

def export_survey_to_excel(data):
    if not data:
        print("⚠️ Survey database khali hai! Excel nahi ban sakti.")
        return
    
    rows = []
    for survey_id, info in data.items():
        rows.append({
            "Survey ID / Roll No": survey_id,
            "Student Name": info["name"].upper(),
            "Class / Batch": info["class"].upper(),
            "Weak Subject": info["weak_subject"].upper(),
            "Kya Samajh Aa Raha Hai?": info["kya_samajh_aaya"],
            "Kya Samajh NAHI Aa Raha?": info["kya_samajh_nahi_aaya"],
            "Coaching Suggestion": info["suggestion"].upper()
        })
    
    df = pd.DataFrame(rows)
    df.to_excel(SURVEY_EXCEL, index=False)
    print(f"📊 Success! Survey Report Excel file '{SURVEY_EXCEL}' me save ho gayi hai.")

# Initialize Survey Data and Counter
survey_db, next_id = load_survey_data()

while True:
    print("\n================== COACHING STUDENT SURVEY SYSTEM ==================")
    print("1. Collect New Survey (Naya Feedback + Auto ID)")
    print("2. View Student Survey (By Survey ID)")
    print("3. Export Survey Data to Excel (Pura Report)")
    print("4. Exit (System Band)")
    print("====================================================================")
    
    choice = input("Option chuniye (1-4): ").strip()

    if choice == "4":
        save_survey_data(survey_db)
        print("\n✅ Saara survey data safe hai. System band ho raha hai!")
        break

    elif choice == "1":
        print("\n--- COLLECT NEW SURVEY ---")
        name = input("Student Ka Naam: ").strip().lower()
        student_class = input("Class / Batch (e.g., 9th, 10th): ").strip().lower()
        weak_subject = input("Kaun sa subject sabse kamzor (weak) hai?: ").strip().lower()
        
        print("\n--- Feedback Questions ---")
        kya_aaya = input("Coaching me kya-kya acche se samajh AA RAHA HAI?: ").strip()
        kya_nahi_aaya = input("Kaun sa topic ya cheez samajh MEI NAHI AA RAHA HAI?: ").strip()
        suggestion = input("Coaching ko aur accha banane ke liye koi salah/suggestion?: ").strip()
        
        # 🆔 AUTOMATIC ID GENERATION 🆔
        survey_id = f"S-{next_id}"
        next_id += 1
        
        # Saving to survey database
        survey_db[survey_id] = {
            "name": name,
            "class": student_class,
            "weak_subject": weak_subject,
            "kya_samajh_aaya": kya_aaya,
            "kya_samajh_nahi_aaya": kya_nahi_aaya,
            "suggestion": suggestion
        }
        
        save_survey_data(survey_db)  # Turant file me save karein
        print(f"\n✅ Success: Data save ho gaya hai!")
        print(f"🆔 Generated Survey ID / Roll No: {survey_id}")

    elif choice == "2":
        print("\n--- VIEW STUDENT SURVEY ---")
        search_id = input("Enter Survey ID (e.g., S-1001): ").strip().upper()
        
        # Agar user sirf number dale (like 1001), toh 'S-' automatic laga dega
        if not search_id.startswith("S-"):
            search_id = f"S-{search_id}"
            
        if search_id in survey_db:
            data = survey_db[search_id]
            print("\n==================================================")
            print(f"SURVEY ID : {search_id} | NAME: {data['name'].upper()} | CLASS: {data['class'].upper()}")
            print(f"📚 Weak Subject: {data['weak_subject'].upper()}")
            print("--------------------------------------------------")
            print(f"👍 Kya samajh aa raha hai:\n   {data['kya_samajh_aaya']}")
            print("--------------------------------------------------")
            print(f"👎 Kya samajh nahi aa raha:\n   {data['kya_samajh_nahi_aaya']}")
            print("--------------------------------------------------")
            print(f"💡 Suggestion/Improvement Point:\n   {data['suggestion']}")
            print("==================================================")
        else:
            print("❌ Is Survey ID ka koi data nahi mila!")

    elif choice == "3":
        export_survey_to_excel(survey_db)
        
    else:
        print("Galat option! Sahi number chuniye.")
```
