```python
import math
def sigmoid(x):
    return 1/(1+math.exp(-x))

x1 = 0.5
x2 = 0.8
x3 = 0.2

w1 = 0.4
w2 = 0.7
w3 = -0.5

b = -0.2
z = w1*x1 + w2*x2+w3*x3+b
a = sigmoid(z)
print("weighted input = ",z)
print("Activation =",a)
```

```python
import numpy as np

def sigmoid (x):
    return 1/(1+np.exp(-x))

x = np.array([0.5,0.8,0.2])

w = np.array([0.4,0.7,-0.5])
b = -0.2
z = np.dot(w,x) +b
a = sigmoid(z)

print("z = ", z)
print("activation = ",a)
```
