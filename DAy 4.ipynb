```python
# ATM version 01
print(" ==============WElocme to ATM  Machine================")
balance = 5000;
while True:
    print("--------------Menu----------------------")
    print("1. Check Balance ")
    print("2. Deposit ")
    print("3. Withdraw ")
    print("4. Exit ")
    
    choose = int(input("Enter number: "))
    if choose == 1:
        print("Current Balance :", balance )
    elif choose == 2:
        deposit = int(input("Enter Deposit Amount$ :"))
        if deposit > 0:
            balance += deposit
            print("Current Balance :", balance )
        else:
            print("Invalid Deposit Amount")
    elif choose ==3:
        withdraw = int(input("Enter Withdraw Amount$: "))
        if withdraw <= balance:
            balance = balance - withdraw
            print("Current Balance :",balance)
        else:
            print("Insufficient Balance ")
    elif choose == 4:
        print("Thank you")
        break
    else:
        print("Invild option")
```
