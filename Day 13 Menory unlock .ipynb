```python
# problem 01
expenses = [500, 1200, 300, 2000, 1500, 700, 2500]
largest = expenses[0] 
for expense in expenses :
    if expense > largest:
        largest = expense 
print(largest) 
    
```

```python
#problem 02
numbers = [3,7,2,7,9,3,9,9]
frequency  = {}
for num in numbers:
    if num not in  frequency:
        frequency[num] = 1 
    else:
        frequency[num] +=1
print(frequency)
    
```
