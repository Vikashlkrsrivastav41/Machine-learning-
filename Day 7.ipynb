```python
class BankAccount:
    def __init__(self,name,balance):
        self.name = name
        self.balance = balance 
    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self,amount):
        if amount <= self.balance:
            self.balance = self.balance - amount

            print("Withdraw Successful")
        else:
            print("Insufficient Balance")
    def check_balance(self):
        print("current Balance =", self.balance)

acc = BankAccount("Vikash", 5000)
acc.deposit(1000)
acc.withdraw(2000)
acc.check_balance()
```

```python
class Student:
    def __init__(self,name, roll,marks):
        self.name = name
        self.roll = roll
        self.marks = marks
    def show(self):
        print("Name :" , self.name)
        print("Roll :" , self.roll)
        print("Marks:", self.marks)
    def update_marks(self,new_marks):
        self.marks = new_marks
        print("Marks Updated:")
    def is_pass(self):
        if self.marks>= 33:
            print("Pass")
        else:
            print("Fail")
    def grade(self):
        if self.marks >= 90:
            print("Grade A")
        elif self.marks >= 75:
            print("Grade B ")
        elif self.marks >= 60:
            print("Grade C")
        else:
            print("Grade D")
        

student1 = Student("Vikash", 101,85)
student1.show()
student1.grade()
student1.is_pass()
student1.update_marks(95)
student1.show()
student1.grade()
```
