```python
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def check_balance(self):
        print("Balance:",self.__balance)
    def deposit(self,amount):
        self.__balance = self.__balance + amount
    def withdraw(self,amount):
        self.__balance = self.__balance - amount

acc  = BankAccount(5000)
acc.check_balance()
acc.deposit(2000)
acc.check_balance()
acc.withdraw(2000)
acc.check_balance()
```

```python
#Abstraction

from abc import ABC, abstractmethod

class Animal(ABC):
    
    @abstractmethod
    def sound(self):
        pass 

class Animal:
    def sound(self):
        print("Animal Sound")

class Dog(Animal):
    def sound(self):
        print("Bhau Bhua")

animal = Animal()
dog1 = Dog()
animal.sound()
dog1.sound()
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Bhau Bhau")


dog1 = Dog()
dog1.sound()
```
