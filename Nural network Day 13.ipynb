```python
#python
import math
x1 = 4
x2 = 5

w1 = 0.3
w2 = 0.6

b = -1
z = w1*x1+w2*x2+b
a = 1/(1+math.exp(-2))

print("z =", z)
print("activation = ",a)
```

```python
#python
import math
x1 = 2
x2 = 4
x3 = 3

w1 = 0.5
w2 = 0.2
w3 = -0.4

b = 0.5
z = w1*x1+w2*x2+w3*x3+b
a = 1/(1+math.exp(-z))

print("z =", z)
print("activation = ",a)
```
